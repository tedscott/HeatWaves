## * Build and write out the full data set including the LSM

```python
%%time

# takes ~6 min to load

# start with one year and lsm and add a coordinate for the year
year = 1961
input_path = '../../../../Data/ERA5-global/Analysis/New-Fourier/Seasons/'+str(year)+'_summer_winter_days.nc'
ds_base = xr.open_dataset(input_path)
ds_base = ds_base.merge(ds_lsm) # add lsm
#ds_base = ds_base.merge(results_ds) # add coastal flag
ds_base = ds_base.expand_dims("time").assign_coords(time=("time", [dt(year,1,1)]))

# loop thru remaining years and also add time coord
input_years = np.arange(1962,2026,1)

for year in input_years:
    input_path = '../../../../Data/ERA5-global/Analysis/New-Fourier/Seasons/'+str(year)+'_summer_winter_days.nc'
    ds_i = xr.open_dataset(input_path)
    ds_i = ds_i.merge(ds_lsm)
    #ds_i = ds_i.merge(results_ds)
    ds_i = ds_i.expand_dims("time").assign_coords(time=("time", [dt(year,1,1)]))
    ds_base = ds_base.merge(ds_i)

ds_base
```

```python
# update attributes
ds_base.FirstWLD.attrs["long_name"] = "First day with mean T2m at or below winter threshold for the year (DOY in [1,365])"
ds_base.FirstWLD.attrs["units"] = "Day of year"

ds_base.LastWLD.attrs["long_name"] = "Last day with mean T2m at or below winter threshold for the year (DOY in [1,365])"
ds_base.LastWLD.attrs["units"] = "Day of year"

ds_base.FirstSLD.attrs["long_name"] = "First day with mean T2m at or above summer threshold for the year (DOY in [1,365])"
ds_base.FirstSLD.attrs["units"] = "Day of year"

ds_base.LastSLD.attrs["long_name"] = "Last day with mean T2m at or above summer threshold for the year (DOY in [1,365])"
ds_base.LastSLD.attrs["units"] = "Day of year"

ds_base.DayOfTmax.attrs["long_name"] = "Day of max mean T2m during the year where summer is centered"
ds_base.DayOfTmax.attrs["units"] = "Day of year"

ds_base.Tmax.attrs["long_name"] = "Max daily mean T2m during the year where summer is centered"
ds_base.Tmax.attrs["units"] = "Degrees [K]"

ds_base.DayOfTmin.attrs["long_name"] = "Day of min mean T2m during the year where winter is centered"
ds_base.DayOfTmin.attrs["units"] = "Day of year"

ds_base.Tmin.attrs["long_name"] = "Min daily mean T2m during the year where winter is centered"
ds_base.Tmin.attrs["units"] = "Degrees [K]"

ds_base.WinterlikeDays.attrs["long_name"] = "Number of days during the year where winter is centered with mean T2m at or below threshold"
ds_base.WinterlikeDays.attrs["units"] = "Days"

ds_base.SummerlikeDays.attrs["long_name"] = "Number of days during the year where summer is centered with mean T2m at or above threshold"
ds_base.SummerlikeDays.attrs["units"] = "Days"

ds_base.lsm.attrs["long_name"] = "Land-sea mask - a measure of the amount of land in the cell"
ds_base.lsm.attrs["units"] = "(0 - 1)"

ds_base
```

```python
# write out complete dataset with LSM and Coastal flags
output_path = '../../../../Data/ERA5-global/Analysis/New-Fourier/Seasons/1961-2025_ALL_summer_winter_days.nc'
ds_base.to_netcdf(output_path)

```

```python
ds = xr.open_dataset('../../../../Data/ERA5-global/Analysis/New-Fourier/Seasons/1961-2025_ALL_summer_winter_days.nc')
ds
```

```python
# find weights (this is a regular grid so we can use cos(latitude))
weights = np.cos(np.deg2rad(ds.lat))
weights.name = "weights"

# split on land vs ocean
ds_land = ds.where(ds.lsm > 0.5)
ds_ocean = ds.where(ds.lsm <= 0.5)
```

### Split land vs ocean

```python
%%time

# takes 20 sec

# latitude & longitude ranges for NH and SH midlatitudes so they will be easy to adjust
nh_min = 23.5
nh_max = 70
sh_min = -23.5
sh_max = -70

# datasets of weighted averages for the midlatitudes
nh_land_full = ds_land.where(((ds_land.lat >= nh_min) & (ds_land.lat <= nh_max))).weighted(weights).mean(dim=['lat','lon'])

sh_land_full = ds_land.where(((ds_land.lat >= sh_max) & (ds_land.lat <= sh_min))).weighted(weights).mean(dim=['lat','lon'])

nh_ocean_full = ds_ocean.where(((ds_ocean.lat >= nh_min) & (ds_ocean.lat <= nh_max))).weighted(weights).mean(dim=['lat','lon'])

sh_ocean_full = ds_ocean.where(((ds_ocean.lat >= sh_max) & (ds_ocean.lat <= sh_min))).weighted(weights).mean(dim=['lat','lon'])

```

### WLD and SLD 

```python
plt.figure(figsize=(15,5))
plt.subplot(1,2,1)
plt.plot(nh_ocean_full.time.dt.year, nh_ocean_full.WinterlikeDays, color="blue", label="Ocean WLD")
plt.plot(nh_land_full.time.dt.year, nh_land_full.WinterlikeDays, color="orange", label="Land WLD")
plt.plot(nh_ocean_full.time.dt.year, nh_ocean_full.SummerlikeDays, color="lightblue", label="Ocean SLD")
plt.plot(nh_land_full.time.dt.year, nh_land_full.SummerlikeDays, color="red", label="Land SLD")
plt.xlabel("Year")
plt.ylabel("Winter Like Days")
plt.title("NH Midlatitude Summer and Winter Like Days, \nOcean vs Land, 1961-2025 \n[23.5, 70]$^\circ$N")
plt.legend()

plt.subplot(1,2,2)
plt.plot(sh_ocean_full.time.dt.year, sh_ocean_full.WinterlikeDays, color="blue", label="Ocean WLD")
plt.plot(sh_land_full.time.dt.year, sh_land_full.WinterlikeDays, color="orange", label="Land WLD")
plt.plot(sh_ocean_full.time.dt.year, sh_ocean_full.SummerlikeDays, color="lightblue", label="Ocean SLD")
plt.plot(sh_land_full.time.dt.year, sh_land_full.SummerlikeDays, color="red", label="Land SLD")
plt.xlabel("Year")
plt.ylabel("Winter Like Days")
plt.title("SH Midlatitude Summer and Winter Like Days, \nOcean vs Land, 1961-2025 \n[-23.5, -70]$^\circ$S")
plt.legend()

plt.show()
```
